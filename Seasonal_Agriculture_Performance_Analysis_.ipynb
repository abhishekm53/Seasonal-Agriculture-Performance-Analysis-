```python
from scipy import stats
from statsmodels.stats.multitest import multipletests

alpha = 0.05

yield_groups = [g["Yield_Tonnes_Ha"].values for _, g in df.groupby("Season")]

anova = stats.f_oneway(*yield_groups)
kruskal = stats.kruskal(*yield_groups)

print(f"One-way ANOVA: F = {anova.statistic:.4f}, p = {anova.pvalue:.6g}")
print(f"Kruskal-Wallis: H = {kruskal.statistic:.4f}, p = {kruskal.pvalue:.6g}")

# Pairwise Mann-Whitney U tests with Bonferroni correction
season_order = list(df["Season"].drop_duplicates())
pair_results = []

for i in range(len(season_order)):
    for j in range(i+1, len(season_order)):
        s1, s2 = season_order[i], season_order[j]
        x = df.loc[df["Season"] == s1, "Yield_Tonnes_Ha"]
        y = df.loc[df["Season"] == s2, "Yield_Tonnes_Ha"]
        u, p = stats.mannwhitneyu(x, y, alternative="two-sided")
        pair_results.append([s1, s2, u, p])

pair_df = pd.DataFrame(pair_results, columns=["Season_1","Season_2","U_statistic","raw_p"])
pair_df["bonferroni_p"] = multipletests(pair_df["raw_p"], method="bonferroni")[1]
pair_df["significant_at_0.05"] = pair_df["bonferroni_p"] < alpha
display(pair_df)
```

```python
n = len(df)
k = df["Season"].nunique()
epsilon_squared = (kruskal.statistic - k + 1) / (n - k)

print(f"Kruskal-Wallis epsilon-squared effect size: {epsilon_squared:.4f}")

# Chi-square test: season vs irrigation method
contingency = pd.crosstab(df["Season"], df["Irrigation_Method"])
chi2, p_chi, dof, expected = stats.chi2_contingency(contingency)

print(f"Chi-square test (Season × Irrigation Method): χ² = {chi2:.4f}, "
      f"df = {dof}, p = {p_chi:.6g}")
```

##  Outlier and Unusual Pattern Analysis

```python
# IQR-based outlier counts for key performance variables
outlier_vars = [
    "Yield_Tonnes_Ha", "Profit_INR", "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3", "Disease_Pest_Risk_pct"
]

outlier_summary = []
for col in outlier_vars:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    mask = (df[col] < lower) | (df[col] > upper)
    outlier_summary.append([col, lower, upper, int(mask.sum()), mask.mean()*100])

outlier_summary = pd.DataFrame(
    outlier_summary,
    columns=["Variable","Lower_bound","Upper_bound","Outlier_count","Outlier_percent"]
)
display(outlier_summary.round(2))

# Show the most extreme yield observations
display(
    df.nlargest(10, "Yield_Tonnes_Ha")[
        ["Farm_ID","State","District","Crop","Season","Yield_Tonnes_Ha",
         "Production_Tonnes","Water_Efficiency_t_per_1000m3","Profit_INR"]
    ]
)
```

## Season-Wise Key Findings

```python
# Generate evidence-based summary values
season_mean = df.groupby("Season").agg({
    "Yield_Tonnes_Ha":"mean",
    "Production_Tonnes":"mean",
    "Revenue_INR":"mean",
    "Profit_INR":"mean",
    "Water_Efficiency_t_per_1000m3":"mean",
    "Rainfall_mm":"mean",
    "Water_Used_m3":"mean",
    "Disease_Pest_Risk_pct":"mean"
}).round(2)

best_yield_season = season_mean["Yield_Tonnes_Ha"].idxmax()
best_profit_season = season_mean["Profit_INR"].idxmax()
best_efficiency_season = season_mean["Water_Efficiency_t_per_1000m3"].idxmax()
highest_rain_season = season_mean["Rainfall_mm"].idxmax()
highest_risk_season = season_mean["Disease_Pest_Risk_pct"].idxmax()

print("Seasonal performance summary")
display(season_mean)

print(f"Highest average yield: {best_yield_season} ({season_mean.loc[best_yield_season,'Yield_Tonnes_Ha']:.2f} t/ha)")
print(f"Highest average profit: {best_profit_season} (₹{season_mean.loc[best_profit_season,'Profit_INR']:,.2f})")
print(f"Highest water efficiency: {best_efficiency_season} ({season_mean.loc[best_efficiency_season,'Water_Efficiency_t_per_1000m3']:.2f} t/1,000 m³)")
print(f"Highest average rainfall: {highest_rain_season} ({season_mean.loc[highest_rain_season,'Rainfall_mm']:.2f} mm)")
print(f"Highest average disease/pest risk: {highest_risk_season} ({season_mean.loc[highest_risk_season,'Disease_Pest_Risk_pct']:.2f}%)")

print("\nStatistical interpretation:")
if kruskal.pvalue < alpha:
    print("The Kruskal-Wallis test indicates statistically significant differences in yield distributions across seasons.")
else:
    print("The Kruskal-Wallis test does not indicate statistically significant differences in yield distributions across seasons.")

if anova.pvalue < alpha:
    print("The one-way ANOVA also indicates statistically significant differences in mean yield.")
else:
    print("The one-way ANOVA does not indicate statistically significant differences in mean yield; this suggests the seasonal difference should be interpreted with attention to distributional assumptions and non-normality/outliers.")

if p_chi < alpha:
    print("Season and irrigation method are statistically associated in this dataset.")
else:
    print("No statistically significant association was detected between season and irrigation method.")
```

##  Data-Driven Recommendations

Based on the analysis, recommendations should be interpreted as **dataset-supported planning suggestions rather than causal claims**:

1. **Prioritize seasonal planning:** Use the observed seasonal yield, profit, rainfall and risk differences to plan crop/resource allocation.
2. **Improve water management:** Compare water use with water-efficiency results and investigate practices used by the highest-efficiency farms.
3. **Monitor high-risk seasons:** The season with the highest observed disease/pest risk deserves closer pest surveillance and preventive planning.
4. **Use crop-season combinations:** Crop performance changes across seasons, so crop selection should consider the observed crop × season yield matrix rather than using one crop strategy for every season.
5. **Investigate regional differences:** State/district-level variation suggests that recommendations may need to be localized.
6. **Validate unusual observations:** Extreme yield, profit, water use or water-efficiency values should be reviewed before making operational decisions.
7. **Avoid causal conclusions from correlation:** Strong correlations identify useful relationships for further study but do not prove that one variable causes another.

The analysis provides a structured view of seasonal agricultural performance across environmental conditions, resource usage, production and economic outcomes. The descriptive analysis identifies how Kharif, Rabi and Zaid differ, while crop-, state- and irrigation-level comparisons reveal additional variation within seasons.

The statistical analysis adds evidence to the visual findings. Correlation analysis highlights variables that move together with yield, while outlier analysis identifies observations requiring further investigation.

Overall, the dataset supports **season-aware agricultural planning**, particularly around crop selection, water efficiency, resource allocation, regional targeting and disease/pest monitoring. These recommendations should be validated with agronomic knowledge and additional field data before being used for high-stakes decisions.
